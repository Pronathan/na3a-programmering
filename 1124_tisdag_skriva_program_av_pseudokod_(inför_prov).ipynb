```python
# program som skriver ut alla tal på en linje tills användaren skriver 0
tal = ""

while True:
  inläsning = input("Skriv ett tal: ")
  if inläsning != "0":
    tal += inläsning
  elif inläsning == "0":
    break

print(tal)
```
