```python
from time import sleep

namn = input("Hej! Vad heter dagens utmanare? ")
print(f"Hej, {namn}")

ålder = int(input("Hur gammal är du? "))
if ålder <= 10:
  print("Du kommer nu få en gåta och har 30 sekunder på dig att lösa den")
  print("Gåta: Vad har fyra ben på dagen, men inga ben alls när den går och lägger sig? ")
  for nedräkning in range (30, -1, -1):
    print(nedräkning)
    sleep(1)
  print("Tiden är ute!")
  svar = input("Vad tror du är svaret? ").lower()
  if svar == "en säng":
    print("Snyggt, du klarade det.")
  else:
    print("Det var tyvärr fel!")

elif ålder > 10 and ålder <= 20:
  print("Du kommer nu få en gåta och har 20 sekunder på dig att lösa den")
  print("Gåta: Vad blir större ju mer du tar bort från det? ")
  for nedräkning in range (20, -1, -1):
    print(nedräkning)
    sleep(1)
  print("Tiden är ute!")
  svar = input("Vad tror du är svaret? ").lower()
  if svar == "ett hål":
    print("Snyggt, du klarade det.")
  else:
    print("Det var tyvärr fel!")

elif ålder >20:
  print("Du kommer nu få en gåta och har 10 sekunder på dig att lösa den")
  print("Gåta: Jag köps för att äta, men ingen äter mig. Vad är jag? ")
  for nedräkning in range (10, -1, -1):
    print(nedräkning)
    sleep(1)
  print("Tiden är ute!")
  svar = input("Vad tror du är svaret? ").lower()
  if svar == "en tallrik" or "bestick" or "sked" or "gaffel":
    print("Snyggt, du klarade det.")
  else:
    print("Det var tyvärr fel!")

```
