```python
from time import sleep
from random import choice

def skriv_långsamt(text):
    for bokstav in text:
        print(bokstav, end = "", flush = True)
        sleep(0.03)
    print()

skriv_långsamt("Hej och välkommen till detta spel som testar dina kunskaper kring logiksekvenser")
namn = input("Vad heter dagens utmanare? ")
print(f"Hej {namn}")

# Meny för svårighetsgrad (väljs bara en gång)
skriv_långsamt("Välj svårighetsgrad:")
print("1. Lätt")
print("2. Medel")
print("3. Svår")

while True:
    svårighet = input("Skriv ditt val (1, 2 eller 3): ").strip()
    if svårighet in ["1", "2", "3"]:
        break
    print("Ogiltigt val, skriv 1 för Lätt, 2 för Medel eller 3 för Svår.")

poäng = 0
senaste_val = None  # Håller koll på senast spelssekvens

# Spelloop – fortsätter tills spelaren når 2 poäng
while poäng < 2:
    # Låt spelaren välja ett NYTT tal vid varje omgång
    while True:
        try:
            valt_tal = int(
                input("\nVälj ett tal mellan 1-20 för nästa utmaning: ")
            )
            if 1 <= valt_tal <= 20:
                break
            else:
                print("Vänligen välj ett tal som är mellan 1 och 20.")
        except ValueError:
            print("Vänligen skriv ett giltigt nummer.")

    if valt_tal % 2 == 0:
        skriv_långsamt("\nOjojoj! Ett svar av det där valet kan få stora följder...")
    else:
        skriv_långsamt(
            "\nOjojoj! Ett svar av det där valet kan få svåra följder, håll i hatten nu!"
        )

    # --- LÄTT ---
    if svårighet == "1":
        alternativ = ["a", "b", "c"] if valt_tal % 2 == 0 else ["d", "e"]

        # Undvik att slumpa samma sekvens som förra gången
        val = choice(alternativ)
        while val == senaste_val and len(alternativ) > 1:
            val = choice(alternativ)
        senaste_val = val

        if val == "a":
            sekvens = "2, 4, 6, 8, ?"  # +2 varje steg
            rätt_svar = 10
        elif val == "b":
            sekvens = "5, 10, 15, 20, ?"  # +5 varje steg
            rätt_svar = 25
        elif val == "c":
            sekvens = "100, 90, 80, 70, ?"  # -10 varje steg
            rätt_svar = 60
        elif val == "d":
            sekvens = "3, 6, 12, 24, ?"  # Dubblering (*2)
            rätt_svar = 48
        elif val == "e":
            sekvens = "1, 4, 7, 10, ?"  # +3 varje steg
            rätt_svar = 13

    # --- MEDEL ---
    elif svårighet == "2":
        alternativ = ["a", "b", "c"] if valt_tal % 2 == 0 else ["d", "e"]

        val = choice(alternativ)
        while val == senaste_val and len(alternativ) > 1:
            val = choice(alternativ)
        senaste_val = val

        if val == "a":
            sekvens = "1, 1, 2, 3, 5, ?"  # Fibonacci
            rätt_svar = 8
        elif val == "b":
            sekvens = "1, 4, 9, 16, ?"  # Kvadrat-tal (1^2, 2^2...)
            rätt_svar = 25
        elif val == "c":
            sekvens = "10, 11, 13, 16, 20, ?"  # Ökar med +1, +2, +3, +4...
            rätt_svar = 25
        elif val == "d":
            sekvens = "5, 10, 8, 13, 11, ?"  # Alternerande (+5, -2)
            rätt_svar = 16
        elif val == "e":
            sekvens = "160, 80, 40, 20, ?"  # Halvering (/2)
            rätt_svar = 10

    # --- SVÅR ---
    elif svårighet == "3":
        alternativ = ["a", "b", "c"] if valt_tal % 2 == 0 else ["d", "e"]

        val = choice(alternativ)
        while val == senaste_val and len(alternativ) > 1:
            val = choice(alternativ)
        senaste_val = val

        if val == "a":
            sekvens = "2, 3, 5, 7, 11, ?"  # Primtal
            rätt_svar = 13
        elif val == "b":
            sekvens = "1, 8, 27, 64, ?"  # Kubik-tal (1^3, 2^3...)
            rätt_svar = 125
        elif val == "c":
            sekvens = "2, 3, 6, 18, 108, ?"  # Multiplicera de två senaste talen
            rätt_svar = 1944
        elif val == "d":
            sekvens = "1, 2, 6, 24, 120, ?"  # Fakultet (*1, *2, *3, *4, *5...)
            rätt_svar = 720
        elif val == "e":
            sekvens = "3, 5, 9, 17, 33, ?"  # Dubbla och dra av 1 (*2 - 1)
            rätt_svar = 65

    skriv_långsamt(f"Vad är nästa tal i sekvensen? {sekvens}")
    gissning = input("Ditt svar: ")

    if gissning.isdigit():
        gissning = int(gissning)
        if gissning == rätt_svar:
            poäng += 1
            skriv_långsamt(
                f"Snyggt jobbat {namn}! Rätt svar. Du har nu {poäng} poäng."
            )
        else:
            poäng = max(0, poäng - 1)
            skriv_långsamt(
                f"Tyvärr fel! Rätt svar var {rätt_svar}. Du har nu {poäng} poäng."
            )
    else:
        skriv_långsamt("Vänligen skriv ett giltigt nummer.")

skriv_långsamt(f"\nGRATTIS {namn.upper()}! Du fick 2 poäng och vann spelet!")
```
