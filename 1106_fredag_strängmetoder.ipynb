```python
def heltal_fraga(fråga):
  while True:
    svar = input(fråga)
    try:
      return int(svar)
    except ValueError:
      print("You need to enter an integer using digits. Spelled-out numbers won't work! Please enter it again.")

CIA_blimps = []

rader_kod = heltal_fraga("How many lines of encrypted code do you have? ")
for index in range(rader_kod):
  print(f"Insert row {index +1}")
  CIA_blimps.append(input())

sökning = "FBI"
blimp = 0

for index,blimp_del in enumerate (CIA_blimps):
  if sökning in blimp_del:
    blimp += 1
    print(f"A CIA blimp was found on row {index + 1}")

if blimp == 0:
  print("HE GOT AWAY!")
```
