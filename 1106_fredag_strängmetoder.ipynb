```python
namn = input("Skriv namnen här med - mellan: ")
namnen = namn.split("-")
initialer = ""

for namn in namnen:
  initialer += namn[0] #tar ut första bokstaven som står i varje namn och sätter in den i variabeln initialer
autori = initialer.upper()

print(autori)

```
