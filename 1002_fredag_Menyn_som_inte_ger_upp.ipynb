```python
from IPython.display import clear_output
saldo = 1000

while True:
  print("Hej och välkommen till denna meny")
  print("1. Visa saldo")
  print("2. Sätta in pengar")
  print("3. Ta ut pengar")
  print("4. Avgift 0.01% på uttag.")
  print("5. Var 10.de körning, avgift 29 kr")
  print("6.Inte övertrassera")

  val = input("Välj ett av alternativen (1-5): ")
  if val == "1":
    print(f"Ditt saldo är {saldo}")
  elif val == "2":
    sätta_in = int(input("Hur mycket vill du sätta in? "))
    saldo += sätta_in
    print(saldo)
  elif val == "3":
    ta_ut = int(input("Hur mycket vill du ta ut? "))
    saldo -= ta_ut
    print(saldo)
  elif val == "4":
    ta_ut = int(input("Hur mycket vill du ta ut? "))
    saldo -= ta_ut
    saldo -= 0.0001 * ta_ut
    print(saldo)
  elif val == "5":
    break
  #elif val == "6":




```
