# Textäventyret - Discväljaren

## Vad berättelsen handlar om

Programmet bygger på att användaren får svara på en rad relevanta frågor kring discgolf. Baserat på vad användaren svara om ex. hur länge man spelat, om man kastar med höger eller vänster arm m.m, så gör programmet en rekommendation av lämplig disc för användaren i just den situation som användaren beskrivit.

## Vägvalen

Jag har väldigt många vägval med de viktigaste är:

start = input("Vill du starta programmet? [ja/nej] ")

Frågan ovan leder till att programet antingen startar igång man svarar ja, eller avslutas om man svarar nej.

kastlängd = heltal_fraga("Hur långt kan du kasta en vanlig kastputter [antal meter]? ")

Frågan ovan leder till att användaren klassas in i en viss grupp: Nybörjare, Amatör, Duktig amatör eller elit nivå.

Finns en rad ytterligare frågor men dessa är starten som kommer avgöra vilken huvudgren man hamnar i.

## Det som var svårast
Det svåraste var att få till loopen där man i slutet först får frågan:

igen = input("Vill du ha en ny rekommendation? [ja/nej]: ")

Därefter ska man antingen få möjlighet till en ny rekommendation eller svara på en utvärdering om man svarar nej. Det svåra med detta var att få till indraget rätt, för det vart flertalet gånger så att programmet startade om loopen utan att man först fick svara på om man ens ville få en ny rekommendation. En annan ganska svår grej var att fixa så att texten skrivs bokstav för bokstav samt krashsäkringen. Det krävdes en del undersökningar för att få till.

## Om jag hade mer tid
Om jag hade mer tid så skulle jag ha fixat så att man bara kunde lägga in sitt namn och därefter få in automatiskt vilken info som finns om en själv som användare. Alltså hur långt man kastar en putter o.s.v. Därefter infon om ett hål som man är på just nu samt info om väderförhållandena (vind, väder m.m.). Sen ska man då alltså få en specifik, personlig rekommendation för just den här situationen. Då skulle man faktiskt få en rekommendation som antagligen funkar, för i nuläget tar ju inte alls programmet hänsyn till vind förhållanden som motvind m.m.
Jag skulle även fixat så att man kunde skriva in vilka discar man själv har tillgång till och att programmet endast rekommenderar discar som man faktiskt har tillgång till. Just nu räknar ju nämligen typ programmet med att man har alla discar som finns i hela världen och det är ju inte verklighetstroget. Programmet kan alltså i nuläget typ bara användas till att rekommendera vilken disc man ska köpa för ett visst typ av flygmönster som man saknat i bagen. Inte vilken disc man faktiskt ska kasta på det här hålet, för det är ju långt ifrån säkert att man har den discen som man blir rekommenderad.
