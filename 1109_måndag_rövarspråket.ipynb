```python
original_ord = input("Vad vill du översätta?")
vokaler = ["a","e","o","i","u","å","ä","ö"]
kosonanter = ["b", "c", "d", "f", "g", "h", "j", "k", "l", "m", "n", "p", "q", "r", "s", "t", "v", "w", "x", "z"]
översättning = ""

for letter in original_ord:
  if letter in vokaler:
    översättning += letter
  else:
    översättning += letter + "o" + letter.lower()

print(översättning)
```

```python
from translate import Translator
original_språk = input("Vilket språk önskar du översätta från? Skriv med bara gemener! ")
förkortning_os = original_språk[0] + original_språk[1]
till_språk = input("Vilket språk önskar du översätta till? Skriv med bara gemener! ")
förkortning_ts = till_språk[0] + till_språk[1]

translator = Translator(from_lang=förkortning_os, to_lang=förkortning_ts)

ord_att_översätta = input("Skriv det du önskar översätta: ")
översatt_ord = translator.translate(ord_att_översätta)

print(f"Översättning: {översatt_ord}")
```
