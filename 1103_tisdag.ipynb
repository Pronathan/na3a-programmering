```python
import random

def heltal_fraga(fråga):
  while True:
    svar = input(fråga)
    try:
      return int(svar)
    except ValueError:
      print("Ogiligt svar, du måste ange ett heltal!")

omgångar = 0
vinster = 0
random_number = random.randint(1, 10)
while omgångar < 3 or spela_igen == "ja":
  guess = heltal_fraga("Gissa ett heltal mellan 1 - 10: ")

  if guess < random_number:
    print("Du gissade för lågt.")
    omgångar += 1
  elif guess > random_number:
    print("Du gissade för högt.")
    omgångar += 1
  else:
    print("Rätt!")
    omgångar += 1
    vinster += 1


  if omgångar % 3 == 0:
    print(f"Du vann {vinster} {'omgång' if vinster == 1 else 'omgångar'}")
    spela_igen = input("Vill du köra en till omgång? [ja/nej] ").lower()
```
