```python
antal_spelare = int(input("Hur många spelare finns det? "))
exclusive = antal_spelare + 1

antal_motståndare = int(input("Hur många motståndare finns det? "))
andra_exclusive = antal_motståndare + 1

for spelare in range(1, exclusive):
  print(spelare)
  for motståndare in range(1, andra_exclusive):
    print(motståndare)
```
