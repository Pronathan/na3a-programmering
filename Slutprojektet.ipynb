```python
import math
import matplotlib.pyplot as plt

# konstanter, motsvarar kolumn J i mitt kalkylark. OBS! Namngivna med VERSALER eftersom de är konstanter
STEGLÄNGD = 0.02
MASSA = 0.175
LUFTENS_DENSITET = 1.225
AREA_DISC = 0.035
DRAGKOEFFICIENT = 0.08
LYFTKOEFFICIENT = 0.15
TYNGDACCELERATION = 9.82

# startvärden, motsvarar rad 2 i mitt kalkylark
x = 0
y = 1.5
kraft = float(input("Hur hårt vill du kasta (t.ex. 15-30 m/s)? "))
vinkel_grader = float(input("Vilken kastvinkel vill du ha (t.ex. 5-25 grader)? "))
vinkel_rad = math.radians(vinkel_grader)
vx = kraft * math.cos(vinkel_rad)
vy = kraft * math.sin(vinkel_rad)

# skapar listor för att spara punkterna som kommer ut
x_lista = []
y_lista = []

# loop som kör tills discen landar på marken
while y>=0:
  # räkna ut total hastighet
  vtot = math.sqrt(vx**2 + vy**2)

  # räkna ut accerleration i x-led och y-led
  ax = -(0.5*LUFTENS_DENSITET*DRAGKOEFFICIENT*AREA_DISC*vtot*vx)/MASSA
  ay = -(0.5*LUFTENS_DENSITET*DRAGKOEFFICIENT*AREA_DISC*vtot*vy)/MASSA + (0.5*LUFTENS_DENSITET*LYFTKOEFFICIENT*AREA_DISC*(vx**2))/MASSA - TYNGDACCELERATION

  # uppdatera position
  x += vx*STEGLÄNGD
  y += vy*STEGLÄNGD

  # uppdatera hastighet
  vx = vx + ax*STEGLÄNGD
  vy = vy + ay*STEGLÄNGD

  # spara de nya koordinaterna i våra listor
  x_lista.append(x)
  y_lista.append(y)

# skriv ut resultatet och rita grafen
print(f"Discen landade efter {round(x, 2)} meter!") # round(x, 2) gör att endast två decimaler skrivs ut. Annars skrivs alla decimaler ut.

plt.figure(figsize=(12, 3)) #Gördiagramrutan mer bred och låg (12 tum bred och 3 tum hög).

plt.plot(x_lista, y_lista, label= "Discens kastbana")
plt.axhline(0, color="green", linestyle="--", label = "Marken") #Ritar en grön linje för marken
plt.axis("equal") #Gör så att diagrammet ser ut mer som kastet ser ut i verkligheten, utan att axlarna förvrängs. (1m i kastet = 1 steg i axeln)
plt.title("Simulering av discgolfkast")
plt.xlabel("Längd (m)")
plt.ylabel("Höjd (m)")
plt.grid(True)
plt.legend()
plt.show()
```
