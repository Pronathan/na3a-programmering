```python
import math
import matplotlib.pyplot as plt

# konstanter, motsvarar kolumn J i mitt kalkylark
steglängd = 0.02
massa = 0.175
luftens_densitet = 1.225
area = 0.035
dragkoefficient = 0.08
lyftkoefficient = 0.15
tyngdacceleration = 9.82

# startvärden, motsvarar rad 2 i mitt kalkylark
x = 0
y = 1.5
kraft = float(input("Hur hårt vill du kasta (t.ex. 15-30 m/s)? "))
vinkel_grader = float(input("Vilken kastvinkel vill du ha (t.ex. 5-25 grader)? "))
vinkel_rad = math.radians(vinkel_grader)
vx = kraft * math.cos(vinkel_rad)
vy = kraft * math.sin(vinkel_rad)

# skapar listor för att spara punkterna som kommer ut
x_lista = []
y_lista = []

# loop som kör tills discen landar på marken
while y>=0:
  # räkna ut total hastighet
  vtot = math.sqrt(vx**2 + vy**2)

  # räkna ut accerleration i x-led och y-led
  ax = -(0.5*luftens_densitet*dragkoefficient*area*vtot*vx)/massa
  ay = -(0.5*luftens_densitet*dragkoefficient*area*vtot*vy)/massa + (0.5*luftens_densitet*lyftkoefficient*area*(vx**2))/massa - tyngdacceleration

  # uppdatera position
  x += vx*steglängd
  y += vy*steglängd

  # uppdatera hastighet
  vx = vx + ax*steglängd
  vy = vy + ay*steglängd

  # spara de nya koordinaterna i våra listor
  x_lista.append(x)
  y_lista.append(y)

# skriv ut resultatet och rita grafen
print(f"Discen landade efter {round(x, 2)} meter!") # (x: .2f) gör att endast två decimaler skrivs ut. Annars skrivs alla ut.

plt.plot(x_lista, y_lista, label= "Discens kastbana")
plt.axhline(0, color="green", linestyle="--", label = "Marken") #Ritar en grön linje för marken
plt.title("Simulering av discgolfkast")
plt.xlabel("Längd (m)")
plt.ylabel("Höjd (m)")
plt.grid(True)
plt.legend()
plt.show()
```
