```python
alfabet = "abcdefghijklmnopqrstuvwxyzåäö"
meddelande = input("Skriv det krypterade ordet: ")

for nyckel in range(1, len(alfabet) + 1):
  test_text = ""
  for bokstav in meddelande:
      pos = alfabet.index(bokstav)
      pos = (pos - nyckel) % len(alfabet)
      test_text += alfabet[pos]
  print(f"nyckel: {nyckel}, text: {test_text}")
```
