```python
A = 7
P = 13
axel_sålt = float(input("Hur många äpplen säljer Axel? "))
axel_förtjänst = (A * axel_sålt)
petra_sålt = float(input("Hur många päron säljer Petra? "))
petra_förtjänst = (P * petra_sålt)

if axel_förtjänst > petra_förtjänst:
  print("Axel har tjänat mer pengar och har därmed vunnit")
elif axel_förtjänst == petra_förtjänst:
  print("De båda har tjänat lika mycket pengar, så ingen av dem vann")
elif petra_förtjänst > axel_förtjänst:
  print("Petra har tjänat mer pengar och har därmed vunnit")
```
